* .groupby(): https://datadoctorblog.com/2023/04/28/Py-Basic-Pandas-Groupby/

```python
df = pd.read_csv("diamonds.csv")
df.head(2)
```

```python
df.groupby("cut")["x"].max()
```

```python
df_g1 = df.groupby("cut")["carat"].max().reset_index()
df_g1
```

```python
df_g2 = df.groupby(["cut", "color"])["carat"].max().reset_index()
df_g2.head()
```

```python
df_g3 = df.groupby("cut")[["carat", "price"]].min().reset_index()
df_g3.head()
```

```python
df_g4 = df.groupby("cut")["price"].agg(["min", "max"]).reset_index()
df_g4["range"] = df_g4["max"] - df_g4["min"]
df_g4.head()
```

```python
df_g5 = df.groupby("cut")[["carat", "price"]].agg({"carat": "min",
                                                   "price": "max"}).reset_index()
df_g5.head() # 시험 기출 유형 🎯🎯
```

```python
df = pd.read_csv("bike.csv")
df_dt2 = df[["datetime"]].copy()
# 단일 column 또는 row 추출 시 series 객체가 아닌 dataframe 객체를 유지 하고 싶다면, 리스트를 추가로 써야 함
# .copy() 는 특히 시험버전의 경우 깊은 복사를 기본적으로 지원하는 경우가 적어 안전한 코딩을 위해서는 필터링 등등 시
# 사용하는 것이 좋음.
```

```python
df_dt2["datetime"] = pd.to_datetime(df_dt2["datetime"])
```

```python
df_dt2["date"  ] = df_dt2["datetime"].dt.date
df_dt2["yyyymm"] = df_dt2["datetime"].dt.strftime("%Y-%m") # 기출 🎯
df_dt2["year"  ] = df_dt2["datetime"].dt.year
df_dt2["month" ] = df_dt2["datetime"].dt.month
df_dt2["wday"  ] = df_dt2["datetime"].dt.weekday
df_dt2["wend"  ] = np.where(df_dt2["wday"] >= 5, 1, 0) # 1: 주말(토, 일), 0: 주중
df_dt2["wend"  ] = np.where(df_dt2["wday"].isin([5, 6]), 1, 0)
df_dt2["wend"  ] = df_dt2["wday"].isin([5, 6]) + 0
df_dt2["hour"  ] = df_dt2["datetime"].dt.hour
df_dt2["dt_s9h"] = df_dt2["datetime"] + pd.to_timedelta(9, unit = "h")
```

```python
df_dt = pd.DataFrame(dict(date = ["2077-08-01", "2077-08-02", "2077-08-03"]))
df_dt["origin"] = "2077-01-01"
df_dt["date"  ] = pd.to_datetime(df_dt["date"])
df_dt["origin"] = pd.to_datetime(df_dt["origin"])
df_dt["diff"  ] = df_dt["date"] - df_dt["origin"]
df_dt["diff_d"] = df_dt["diff"].dt.days
df_dt
```

```python
df_t = pd.DataFrame(dict(date = pd.date_range("2027-01-01", "2027-01-07"),
                         v = [100, 200, 300, np.nan, 500, 600, 700]))
df_t["v_s1"] = df_t["v"].shift(1)
df_t["v_sn1"] = df_t["v"].shift(-1)
df_t["v_fill_1"] = np.where(df_t["v"].isna(),
                            (df_t["v_s1"] + df_t["v_sn1"]) / 2, 
                            df_t["v"])
df_t["v_fill_2"] = df_t["v"].interpolate()
```
