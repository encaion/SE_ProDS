```python
df_m2 = df.melt(id_vars = ["YEAR", "MONTH", "DAY"], var_name = "HOUR", value_name = "LOAD")
df_m2.head()
```

```python
df_m2["h1"] = df_m2["HOUR"].str.replace("X", "").str.replace("HR", "")
df_m2["h2"] = df_m2["HOUR"].str.replace("X|HR", "", regex = True) # X 또는 HR
df_m2["h3"] = df_m2["HOUR"].str.replace("[A-Z]", "", regex = True) # 영문 대문자
df_m2["h4"] = df_m2["HOUR"].str.replace("[^0-9]", "", regex = True) # 숫자를 제외한 나머지
df_m2["h5"] = df_m2["HOUR"].str.extract("([0-9]{1,2})") # 숫자 1~2자리 최초 매칭분 추출.
```

```python
# 11년도 요일별 count의 평균
# 12년도 요일별 count의 평균
# 하나로 합치고 비교해볼까?
df_q = df_bike[["datetime", "count"]].copy()
df_q.head(2)
```

```python
df_q["datetime"] = pd.to_datetime(df_q["datetime"])
df_q["year"] = df_q["datetime"].dt.year
df_q["wday"] = df_q["datetime"].dt.weekday
df_q.head(2)
```

```python
df_y11 = df_q.loc[df_q["year"] == 2011, ]
df_y12 = df_q.loc[df_q["year"] == 2012, ]
len(df_y11), len(df_y12)
```

```python
df_y11_g = df_y11.groupby("wday")["count"].mean().reset_index()
df_y12_g = df_y12.groupby("wday")["count"].mean().reset_index()
df_y11_g = df_y11_g.rename(columns = {"count": "cnt_11"})
df_y12_g = df_y12_g.rename(columns = {"count": "cnt_12"})
df_y11_g.head(2)
```

```python
df_g_join1 = pd.merge(df_y11_g, df_y12_g, on = "wday", how = "inner")
df_g_join1["diff"] = df_g_join1["cnt_12"] - df_g_join1["cnt_11"]
df_g_join1["ratio"] = df_g_join1["cnt_12"] / df_g_join1["cnt_11"]
df_g_join1
```
