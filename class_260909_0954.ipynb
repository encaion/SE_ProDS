```python
~(df_s["v1"] == 200)
```

```python
df_s.loc[~(df_s["v1"] == 200), ]
```

```python
df_s.loc[(df_s["v1"] == 200) & (df_s["v2"] == 600), ]
```

```python
df_s.loc[(df_s["v1"] == 200) | (df_s["v2"] == 600), ]
```

```python
condi_1 = df_s["v1"] == 200
condi_2 = df_s["v2"] == 500
df_s.loc[condi_1 & condi_2, ]
```

```python
df_s.loc[(df_s["shape"] == "A") | (df_s["shape"] == "C"), ]
```

```python
df_s.loc[df_s["shape"].isin(["A", "C"]), ]
```

```python
df_s.loc[~df_s["shape"].isin(["A", "C"]), ]
```
